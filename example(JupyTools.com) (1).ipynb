## База 1 – Фокус: Функции и модули

### Задача
Реализовать функцию `solve_base1(value)`, которая возвращает `-1`, если `value < 0`, иначе `value + 1`.  
Продемонстрировать использование модуля (в рамках одной ячейки эмулируем модуль).

```python
# base1_module.py (эмуляция модуля)
def solve_base1(value: int) -> int:
    """Возвращает -1 для отрицательных value, иначе value + 1."""
    if value < 0:
        return -1
    return value + 1

# Использование и тестирование
def test_base1():
    test_cases = [5, 0, -1, -100, 100]
    for v in test_cases:
        result = solve_base1(v)
        print(f"solve_base1({v}) = {result}")
    # Проверка на граничные значения с assert
    assert solve_base1(5) == 6
    assert solve_base1(0) == 1
    assert solve_base1(-1) == -1
    assert solve_base1(-100) == -1
    print("\nВсе assert'ы Базы 1 пройдены")

test_base1()
```

### Рефлексия Базы 1
**Основной риск:** неверная обработка отрицательных значений.  
**Как закрыт:** явное условие `if value < 0`, тесты покрывают границы.  
**Демонстрация темы:** функция вынесена в отдельный логический блок (модуль), повторно используется.

## База 2 – Фокус: Структура программы

### Задача
Реализовать функцию `solve_base2(value)`, возвращающую `-1` при `value < 0`, иначе `value + 2`.  
Акцент на чёткой структуре программы (разделение на функции, защита входа, комментарии).

```python
# -*- coding: utf-8 -*-
"""
Структура программы:
- Вспомогательная функция проверки
- Основная решающая функция
- Блок тестирования
"""

def is_negative(val: int) -> bool:
    """Возвращает True, если val < 0."""
    return val < 0

def solve_base2(value: int) -> int:
    """Решает задачу согласно шаблону."""
    if is_negative(value):
        return -1
    return value + 2

def run_tests_base2():
    """Функция для запуска проверок."""
    test_data = [10, 0, -5, -1]
    for x in test_data:
        print(f"solve_base2({x}) = {solve_base2(x)}")
    # Asserts для граничных случаев
    assert solve_base2(10) == 12
    assert solve_base2(0) == 2
    assert solve_base2(-5) == -1
    assert solve_base2(-1) == -1
    print("\nВсе assert'ы Базы 2 пройдены")

if __name__ == "__main__":
    # Точка входа (в notebook это не обязательно, но структура показана)
    print("Тестирование решения База 2")
    run_tests_base2()

# Вызов для демонстрации
run_tests_base2()
```

### Рефлексия Базы 2
**Риск:** хаотичное смешение логики, сложность поддержки.  
**Как закрыт:** разделение на функции, защищённая точка входа, документирование.  
**Тема «Структура программы»:** программа организована в логические блоки, легко читается и тестируется.

## База 3 – Фокус: Данные

### Задача
Реализовать функцию `solve_base3(value)`, возвращающую `-1` при `value < 0` или нечисловом входе, иначе `value + 3`.  
Акцент на гибкой обработке различных типов данных.

```python
def solve_base3(value):
    """
    Принимает данные разных типов (int, float, str, None).
    Возвращает -1, если вход нечисловой или отрицательный,
    иначе value + 3 (преобразуя в int).
    """
    if value is None:
        return -1
    if isinstance(value, str):
        try:
            value = int(value)
        except ValueError:
            return -1
    if not isinstance(value, (int, float)):
        return -1
    if value < 0:
        return -1
    return int(value) + 3

# Проверка на различных типах данных (пограничные случаи)
test_inputs = [5, 0, -2, "10", "abc", None, 3.14, [], {}]
for inp in test_inputs:
    print(f"solve_base3({repr(inp)}) = {solve_base3(inp)}")

# Дополнительные assert'ы для границ
assert solve_base3(5) == 8
assert solve_base3(0) == 3
assert solve_base3(-2) == -1
assert solve_base3("10") == 13
assert solve_base3("abc") == -1
assert solve_base3(None) == -1
assert solve_base3([]) == -1
print("\nВсе assert'ы Базы 3 пройдены")
```

### Рефлексия Базы 3
**Риск:** падение программы при `None` или нечисловых данных.  
**Как закрыт:** проверка типа, преобразование строк, защита от пустых коллекций.  
**Тема «Данные»:** функция корректно работает с числами, строками, `None` и неподдерживаемыми типами.

## Стандарт 1 – Фокус: Функции и модули (углублённый уровень)

### Задача
Реализовать функцию `solve_standard1(value)` по тому же шаблону (value < 0 → -1, иначе value + 1), но с использованием **пакетной структуры, документирования и юнит-тестов**.  
В среде Jupyter Notebook эмулируем модуль и запускаем `unittest`.

```python
# core.py - эмуляция модуля
"""Базовый модуль с решающей функцией."""
def solve_standard1(value: int) -> int:
    """
    Стандартное решение.
    
    Args:
        value (int): Входное целое число.
    
    Returns:
        int: -1 если value < 0, иначе value + 1.
    """
    if value < 0:
        return -1
    return value + 1
```

```python
# Юнит-тесты с использованием unittest в notebook (без выхода из процесса)
import unittest
import sys
from io import StringIO

class TestSolveStandard1(unittest.TestCase):
    def test_positive(self):
        self.assertEqual(solve_standard1(5), 6)
        self.assertEqual(solve_standard1(0), 1)
    
    def test_negative(self):
        self.assertEqual(solve_standard1(-1), -1)
        self.assertEqual(solve_standard1(-100), -1)
    
    def test_large_positive(self):
        self.assertEqual(solve_standard1(1000), 1001)

# Запуск тестов с захватом вывода
runner = unittest.TextTestRunner(stream=StringIO(), verbosity=2)
suite = unittest.TestLoader().loadTestsFromTestCase(TestSolveStandard1)
result = runner.run(suite)

# Вывод результатов
print("Результаты unittest:")
for test, output in zip(suite, result.failures + result.errors):
    print(f"Проблема: {test}")
if result.wasSuccessful():
    print("Все тесты успешно пройдены!")
else:
    print(f"Пройдено: {result.testsRun - len(result.failures) - len(result.errors)} из {result.testsRun}")
```

### Рефлексия Стандарт 1
**Риск:** при изменении кода легко внести ошибку.  
**Как закрыт:** автоматические тесты и модульная организация.  
**Уровень «Стандарт»:** продемонстрирована профессиональная практика – пакет (эмуляция модуля), документация, unittest, повторное использование.
